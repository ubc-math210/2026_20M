## Example

### Quadratic Formula

The roots of $p(x) = ax^2 + bx + c$ are given by the quadratic formula

$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

Compute the roots of $x^2 + x - 1$.

```python
(-1 + (1**2 - 4*(1)*(-1))**0.5)/(2*1)
```

```python
(-1 - (1**2 - 4*(1)*(-1))**0.5)/(2*1)
```

## Taylor Series

The Taylor series of $e^x$ centered at $x=0$ is

$$
e^x = \sum_{k=0}^{\infty} \frac{x^k}{k!}
$$

Approximate $e$ using the first 6 terms of the Taylor series (evaluated at $x=1$).

```python
1 + 1 + 1/2 + 1/(3*2) + 1/(4*3*2) + 1/(5*4*3*2)
```

The Taylor series of $\ln(1 + x)$ centered at $x=0$ is

$$
\ln(1 + x) = \sum_{k=1}^{\infty} (-1)^{k+1} \frac{x^k}{k}
$$

Use the first 6 terms to approximate $\ln(1/2)$.

```python
(-1/2)**1/1 - (-1/2)**2/2 + (-1/2)**3/3 - (-1/2)**4/4 + (-1/2)**5/5 - (-1/2)**6/6
```

### Limits

Consider the limit

$$
e = \lim_{n \to \infty} \left( 1 + \frac{1}{n} \right)^n
$$

Compute the value for $n=1000$.

```python
(1 + 1/1000)**1000
```

## Variables

Python variables allow us to store and reuse data to write comptuations more efficiently.

```python
a = 2
b = 5
c = -4
r1 = (-b + (b**2 - 4*a*c)**0.5)/(2*a)
r2 = (-b - (b**2 - 4*a*c)**0.5)/(2*a)
print(r1,r2)
```

```python
x = -1
y = 1 + x + x**2/2 + x**3/(3*2) + x**4/(4*3*2) + x**5/(5*4*3*2)
print(y)
```

```python
n = 100000
L = (1 + 1/n)**n
print(L)
```
