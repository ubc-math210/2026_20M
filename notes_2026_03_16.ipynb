```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as spi
```

## `odeint`

```python
# mx'' + cx' + kx = 0 , x(0) = x0 , x'(0) = 0
# u0 = x; u1 = x'; u = [u0,u1];
# u0' = u1
# u1' = (-c*u1-k*u0)/m
m = 1; c = 1; k = 10; tf = 5; x0 = 5;
f = lambda u,t: [u[1] , (-c*u[1] - k*u[0])/m]
uinitial = [x0,0]
t = np.linspace(0,tf,200)
U = spi.odeint(f,uinitial,t)
```

```python
U[:5,:]
```

```python
plt.plot(t,U[:,0])
plt.show()
```
