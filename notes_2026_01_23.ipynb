## `for` loops

A `for` loop allows us to execute a block of code repeatedly with an index that updates over every entry of a sequence. The syntax is:

```
for item in sequence:
    Python code
    Python code
    Python code
```
where:
* `for` is the keyword that starts the construction
* `item` is a variable name
* `in` is a keyword and `sequence` is a list or range (or any other sequence)
* `:` ends the `for` statement
* block of code is indented 4 spaces
* run the block for `item` assigned each value in `sequence`

````python
for n in range(5):
    print(n**2)
````

## Example: Factorial

The value $n!$ is $n(n-1)(n-2) \cdots (2)(1)$. We can write this as a recursive sequence: $x_0 = 1$, $x_k = k x_{k-1}$ for $k=1,\dots,n$.

````python
n = 5
x = 1
for k in range(1,n+1):
    x = k*x
    print(x)
````

Write a function called `factorial` which takes an integer $n$ and returns $n!$. Recall $0! = 1$.

````python
def factorial(n):
    x = 1
    for k in range(1,n+1):
        x = k*x
    return x
````

````python
print(factorial(0),1)
````

````python
print(factorial(1),1)
````

````python
print(factorial(5),120)
````

````python
print(factorial(10),2*3*4*5*6*7*8*9*10)
````

Write a function called `exp` which takes inputs `x` and `N` and returns the partial sum of the Taylor series of $e^x$:
$$
e^x \approx \sum_{k=0}^N \frac{x^k}{k!}
$$

````python
def exp(x,N):
    terms = [x**k/factorial(k) for k in range(0,N+1)]
    value = sum(terms)
    return value
````

````python
print(exp(0,12),1.)
````

````python
print(exp(1,10),2.718281828)
````

````python
print(exp(2,20),7.389056099)
````

## Example: Recursive Sequence

Write a function `aseq` which takes inputs `a` and `N` and returns $x_N$ defined by the recursive sequence: $x_0 = a$, $x_{k} = \sqrt{2 + x_{k-1}}$.

````python
def aseq(a,N):
    x = a
    for k in range(1,N+1):
        x = (2 + x)**0.5
    return x
````

````python
print(aseq(0,1),2**0.5)
````

````python
#x0 = 1
#x1 = (2 + x0)**0.5 = 3**0.5
#x2 = (2 + x1)**0.5 = (2 + 3**0.5)**0.5
print(aseq(1,2),(2 + 3**0.5)**0.5)
````

## Remove Zeros

Write a function called `remove0s` which takes a list of numbers `p` and returns the same list but with all the zeros (from the end) removed. For example, if `p = [1,2,3,0,0,0]` the function returns `[1,2,3]`.

Notes:
* If `p` is all 0s then return `[0]`.
* Use the keyword `break` to exit a loop before the loop is complete

````python
def remove0s(p):
    for k in range(len(p)):
        if p[-k-1] == 0:
            print("Remove p[-k-1] for k =",k)
        if p[-1] != 0:
            break
    return [1,1,1]
````

````python
print(remove0s([1,2,3,0,0]),[1,2,3])
````

````python
print(remove0s([1,2,3]),[1,2,3])
````

````python
print(remove0s([1,-1,1,-1,1,0]),[1,-1,1,-1,1])
````
