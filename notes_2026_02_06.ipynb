## Newton's Method

Let $f(x)$ be a differentiable function, let $r$ be a root of $f(x)$ and let $x_0$ be some value near $r$. Construct the tangent line to $y=f(x)$ at $x_0$
$$
y = f'(x_0)(x - x_0) + f(x_0)
$$
Idea: If $x_0$ is near the root $r$ then the $x$-intercept of the tangent should be close to $r$. Find the $x$-intercept:
$$
0 = f'(x_0)(x - x_0) + f(x_0)
\Rightarrow
x = x_0 - \frac{f(x_0)}{f'(x_0)}
$$
Repeat! Newton's method applied to $f(x)$ with initial value $x_0$ is the recursive sequence
$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

Write a function called `newton` which takes functions `f` and `df`, number `x0` and integer `N`, and return $x_N$ the $N$th value in the sequence given by Newton's method.

```python
def newton(f,df,x0,N):
    xn = x0
    for n in range(N):
        xn = xn - f(xn)/df(xn)
    return xn
```

```python
f = lambda x: x**2 - 2
df = lambda x: 2*x
x0 = 1.414
N = 10
r = newton(f,df,x0,N)
print(r,2**0.5)
```

Compare to secant and bisection methods:

```python
def secant(f,a,b,N):
    if f(a)*f(b) >= 0:
        print("Initial interval does not satisfy f(a)f(b) < 0.")
        return None
    a_n = a
    b_n = b
    for n in range(1,N+1):
        f_a_n = f(a_n)
        f_b_n = f(b_n)
        m_n = a_n - f_a_n*(b_n - a_n)/(f_b_n - f_a_n)
        f_m_n = f(m_n)
        if f_a_n*f_m_n < 0:
            b_n = m_n
        elif f_b_n*f_m_n < 0:
            a_n = m_n
        elif f_m_n == 0:
            print("Found exact solution.")
            return m_n
        else:
            print("Secant method fails.")
            return None
    return a_n - f(a_n)*(b_n - a_n)/(f(b_n) - f(a_n))
```

```python
def bisection(f,a,b,N):
    if f(a)*f(b) >= 0:
        print("Initial interval does not satisfy f(a)f(b) < 0.")
        return None
    a_n = a
    b_n = b
    for n in range(1,N+1):
        m_n = (a_n + b_n)/2
        f_m_n = f(m_n)
        if f(a_n)*f_m_n < 0:
            b_n = m_n
        elif f(b_n)*f_m_n < 0:
            a_n = m_n
        elif f_m_n == 0:
            print("Found exact solution.")
            return m_n
        else:
            print("Bisection method fails.")
            return None
    return (a_n + b_n)/2
```

```python
f = lambda x: x**2 - 2
df = lambda x: 2*x
x0 = 2; N = 3; a = 1; b = 2;
r = 2**0.5
rnew = newton(f,df,x0,N)
rsec = secant(f,a,b,N)
rbis = bisection(f,a,b,N)
print("Newton's Method Error:",abs(r - rnew))
print("Secant Method Error:",abs(r - rsec))
print("Bisection Method Error:",abs(r - rbis))
```

## Polynomial Example

Write a function called `poly_root` which takes list `p` (representing a polynomial), number `x0`, and integer `N`, and return $x_N$ the value after $N$ iterations of Newton's method.

```python
def poly_eval(p,a):
    terms = [p[n]*a**n for n in range(len(p))]
    return sum(terms)
    
def poly_diff(p):
    if len(p) == 1:
        return [0]
    else:
        return [n*p[n] for n in range(1,len(p))]

def poly_root(p,x0,N):
    f = lambda x: poly_eval(p,x)
    df = lambda x: poly_eval(poly_diff(p),x)
    r = newton(f,df,x0,N)
    return r
```

```python
p = [2,0,-1] # p(x) = 2 - x^2
x0 = 1.414
N = 10
r = poly_root(p,x0,N)
print(1.4142135623730951,r)
```

Let $f(x) = x^5 + x^3 + x - 5$.

* How many real roots?
* Where are they?

Compute $f'(x) = 5x^4 + 3x^2 + 1$. Then $f'(x) > 0$ for all $x$ because $x^4 \geq 0$ and $x^2 \geq 0$ for all $x$. Since $f(x)$ is always increasing there is at most 1 root. Plug in $f(0) = -5$ and $f(1) = -2$ and $f(2) = 2^5 + 2^3 + 2 - 5 > 0$. There is a root in $[1,2]$.

```python
p = [-5,1,0,1,0,1]
x0 = 1.5
N = 10
r = poly_root(p,x0,N)
r
```

```python
poly_eval(p,r)
```
