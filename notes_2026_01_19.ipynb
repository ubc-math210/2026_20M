## Polynomial Function

Represent a polynomial $p(x) = c_0  + c_1 x + \cdots + c_d x^d$ as a list of coefficients `p = [c0,c1,...,cd]`.

Write a function called `poly_eval` which takes a Python list of numbers `p` (representing polynomial $p(x)$) and number `a`, and returns
$$
p(a) = c_0 + c_1 a + c_2 a^2 + \cdots + c_d a^d = \sum_{k=0}^d c_k a^k
$$

````python
def poly_eval(p,a):
    d = len(p) - 1
    terms = [p[k]*a**k for k in range(d + 1)]
    pa = sum(terms)
    return pa
````

````python
# p(x) = 1 + x + x^2
p1 = [1,1,1]
a1 = -1
result1 = poly_eval(p1,a1)
print(result1,1)
````

````python
# p(x) = 1 - 4*x^5
p2 = [1,0,0,0,0,-4]
a2 = 1/2
result2 = poly_eval(p2,a2)
print(result2,1 - 4*(1/2)**5)
````

## `if` statements

Execute different blocks of code given some boolean values. The construction is:

```
if condition1:
   Python code block 1
elif condition2:
   Python code block 2
else:
   Python code block 3
```

Let's write an `if` statement to print whether roots of polynomial $p(x) = ax^2 + bx + c$ are real and distinct, real and repeated or complex. Recall the quadratic formula:

$$
r = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

````python
a = 1; b = 0; c = 1;
D = b**2 - 4*a*c

if D > 0:
    print("Roots are real and distinct.")
    print((-b + D**0.5)/(2*a),(-b - D**0.5)/(2*a))
elif D < 0:
    print("Roots are complex.")
    print(-b/(2*a),"+-",(-D)**0.5/(2*a),"j")
else:
    print("Roots are real and repeated.")
    print(-b/(2*a))

print("Done!")
````

## Example

Write a function called `poly_diff` which takes a Python list of numbers `p` (representing a polynomial $p(x)$) and returns the Python list of numbers representing $p'(x)$.

````python
def poly_diff(p):
    if len(p) > 1:
        print("Differentiate p(x)")
        dpdx = [1.0,1.0,1.0]
    else:
        dpdx = [0]
    return dpdx
````

````python
p1 = [1,2,3] # p(x) = 1 + 2x + 3x^2
dp1dx = poly_diff(p1)
print(dp1dx,[2,6]) # p'(x) = 2 + 6x
````

````python
p2 = [0,1,0,0,4] # p(x) = x + 4x^4
dp2dx = poly_diff(p2)
print(dp2dx,[1,0,0,16]) # p'(x) = 1 + 16x^3
````

````python
p3 = [1] # p(x) = 1
dp3dx = poly_diff(p3)
print(dp3dx,[0]) # p'(x) = 0
````
