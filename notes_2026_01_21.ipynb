## Example: Differentiation Polynomials

Represent a polynomial $p(x) = c_0 + c_1 x + \cdots + c_d x^d$ as a list of coefficients `p = [c0,c1,...,cd]`.

Write a function called `poly_diff` which takes a Python list of numbers `p` (representing a polynomial $p(x)$) and returns the Python list of numbers representing $p'(x)$.

$$
p'(x) = c_1 + 2c_2 x + 3 x_3 x^2 + \cdots + d c_d x^{d-1} = \sum_{k=1}^d k c_k x^{k-1}
$$

````python
def poly_diff(p):
    if len(p) > 1:
        d = len(p) - 1
        dpdx = [k*p[k] for k in range(1,d+1)]
    else:
        dpdx = [0]
    return dpdx
````

````python
def poly_diff(p):
    if len(p) > 1:
        return [k*p[k] for k in range(1,len(p))]
    else:
        return [0]
````

````python
p1 = [1,2,3] # p(x) = 1 + 2x + 3x^2
dp1dx = poly_diff(p1)
print(dp1dx,[2,6]) # p'(x) = 2 + 6x
````

````python
p2 = [0,1,0,0,4] # p(x) = x + 4x^4
dp2dx = poly_diff(p2)
print(dp2dx,[1,0,0,16]) # p'(x) = 1 + 16x^3
````

````python
p3 = [1] # p(x) = 1
dp3dx = poly_diff(p3)
print(dp3dx,[0]) # p'(x) = 0
````

## Example: Add Polynomials

Write a function called `poly_add` which takes Python lists of numbers `p` and `q` (representing polynomials $p(x)$ and $q(x)$) and returns the Python list of numbers representing the polynomial $p(x) + q(x)$. We require the last entry in a list representing a polynomial to be nonzero (except if `p = [0]`).

````python
def poly_add(p,q):
    if len(p) == len(q):
        pq = [p[k] + q[k] for k in range(len(p))]
    elif len(p) < len(q):
        pq = [p[k] + q[k] for k in range(len(p))] + q[len(p):]
    elif len(p) > len(q):
        pq = [p[k] + q[k] for k in range(len(q))] + p[len(q):]
    return pq
````

````python
p1 = [1,1,1]
q1 = [1,2,3,4]
pq1 = poly_add(p1,q1)
print(pq1,[2,3,4,4])
````

````python
p2 = [1,0,-1,2,3]
q2 = [1,-2]
pq2 = poly_add(p2,q2)
print(pq2,[2,-2,-1,2,3])
````

````python
p3 = [1,2,-1]
q3 = [1,-2,1]
pq3 = poly_add(p3,q3)
print(pq3,[2])
````

````python
p4 = [1,0,-1,2]
q4 = [1,-2,1,1]
pq4 = poly_add(p4,q4)
print(pq4,[2,-2,0,3])
````

We need a for loop to complete this function to remove extra 0s at the end!

## `for` loops

A `for` loop allows us to run a block of code multiple times using an index at each execution of the block. The syntax is:

```
for item in sequence:
    Python code block
    Python code block
    Python code block
```
where:
* `item` is variable name
* `sequence` is a Python sequence (usually `range(a,b)`)
* run code for each value of `sequence` assigned to `item`

````python
for n in range(5):
    print(n**2)
print("Done!")
````
