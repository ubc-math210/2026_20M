## JupyterLab

## Cells and Modes

There are 2 types of cells: code cells and markdown cells. Markdown cells are for text, markdown, HTML, LaTeX. Code cells are for Python code.

Switch between cell types in the toolbar above or use key board shortcuts in edit mode: `y` for code cell and `m` for markdown.

Switch between edit mode and command mode using `ESC` and `ENTER`.

Run a cell with `SHIFT + ENTER`.

## Markdown

Use `#` for heading 1, `##` for heading 2, etc.

## LaTeX

LaTeX is a development environment for creating scientific documents. Write mathematical formulas using LaTeX syntax in `$ $`.

Use single dollar signs for inline LaTeX: $f(x) = e^{-x^2} \cos(x)$.

Use double dollar signs for display LaTeX (appearing in its own paragraph):

$$
\int_a^b f(x) dx = F(b) - F(a)
$$

From MATH 215 lecture today, a **differential equation** is an equation involving an unknown function $y(t)$ and its derivatives. The **order** of a differential equation is the highest order derivative of the unknown function appearing in the equation. For example, the following is a first order differential equation:

$$
y' + y = \cos(t)
$$
