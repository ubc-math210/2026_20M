## Roots and IVT

A **root** of a function $g(x)$ is a value $r$ such that $g(r) = 0$. We also call a root a **zero** of the function.

What's the connection between fixed points and roots? If $r$ is a root of $g(x)$ then $r$ is a fixed point of $f(x) = x - g(x)$.

Recall the Intermediate Value Theorem:

Let $g(x)$ be continuous on $[a,b]$. If $g(a)g(b) < 0$ then there is some $c \in (a,b)$ such that $g(c) = 0$.

## Bisection Method

The bisection method is a root finding method. It applies IVT over and over again taking smaller and smaller intervals.

* Start with $[a_0,b_0]$ such that $f(a_0) f(b_0) <0$.
* Let $m_0 = (a_0 + b_0)/2$ be the midpoint.
* Then either: (i) $f(a_0)f(m_0) < 0$ or (ii) $f(m_0)f(b_0) < 0$ (or $f(m_0) = 0$).
* Define: (i) $a_1 = a_0$, $b_1 = m_0$ or (ii) $a_1 = m_0$, $b_1 = b_0$ so that $r \in (a_1,b_1)$.
* Repeat!

Write a function called `bisection` which takes function `f`, numbers `a` and `b` and integer `N`, and return the midpoint $m_N$ after $N$ iterations of the bisection mtehod applied to $f(x)$.

```python
def bisection(f,a,b,N):
    if f(a)*f(b) >= 0:
        print("Initial interval does not satisfy f(a)f(b) < 0.")
        return None
    a_n = a
    b_n = b
    for n in range(1,N+1):
        m_n = (a_n + b_n)/2
        f_m_n = f(m_n)
        if f(a_n)*f_m_n < 0:
            b_n = m_n
        elif f(b_n)*f_m_n < 0:
            a_n = m_n
        elif f_m_n == 0:
            print("Found exact solution.")
            return m_n
        else:
            print("Bisection method fails.")
            return None
    return (a_n + b_n)/2
```

```python
f = lambda x: x**3 + x - 1
a = 0
b = 1
N = 10
r = bisection(f,a,b,N)
print(r)
```

It looks like fixed point iteration is converging to $L \approx 4.439$. Prove $|f'(x)| < 1$ for $x$ near $L$.

$$
f'(x) = \frac{1}{2 \sqrt{1 + x}} + \frac{1}{2\sqrt{x}}
$$

Since each denominator is increasing as $x$ is increasing, the value of each term is decreasing as $x$ is increasing.

Look at $f'(1) = \frac{1}{2 \sqrt{2}} + \frac{1}{2} < 1$ and note $f'(x)$ is always positive and decreasing for $x$ increasing from 1. Therefore $|f'(x)| < 1$ for all $x \geq 1$ and so fixed point iteration converges for any $x \geq 1$.
