## Sequences

There are 3 main sequence types: list, tuple, and range. Python treats these types differently and we use them in different ways.

Use square brackets to make a list:

```python
primes = [2,3,5,7,11,13,17,23,29,31,37,41]
```

```python
print(primes)
```

```python
type(primes)
```

A range objects represents a sequence of evenly spaced integers. In particular, `range(a,b,step)` then we get the sequence of integers from `a` to `b` (exclusive) incremented by `step`.

```python
integers = range(1,17,2)
```

```python
type(integers)
```

```python
print(integers)
```

A range object represents a sequence but it only produces the values when you ask for it. For example, it will return each value in a for loop one at a time. Or we can convert it to a list to get all the values at once.

```python
integers_list = list(integers)
```

```python
type(integers_list)
```

```python
print(integers_list)
```

If we don't specify `step` then range increments by 1.

```python
N = 100
integers_N = list(range(1,N+1))
print(integers_N)
```

There is also `tuple`. It behaves almost exactly like a list and so we'll just uses lists. Use round parentheses to create a tuple.

```python
date = (2026,1,12)
```

```python
type(date)
```

```python
print(date)
```

## List Comprehensions

Construct sequences of numbers from a formula using a list comprehension:

```python
seq = [1/n**2 for n in range(1,10)]
print(seq)
```

```python
1/9**2
```

Use a list comprehension to compute
$$
\sum_{n=1}^{100} \frac{1}{n^2}
$$

```python
seq = [1/n**2 for n in range(1,101)]
series = sum(seq)
print(series)
```
