## Fixed Points

Consider the recursive sequence from last class: $x_0 = 1$, $x_{k+1} = \sqrt{2 + x_k}$. Let's use a `for` loop to compute $x_{10}$.

```python
a = 1; N = 10;
xk = a
for k in range(1,N+1):
    xk = (2 + xk)**0.5
print(xk)
```

Note that we can construct a recursive sequence for any function $f(x)$. In particular, $x_0 = a$ and $x_{k+1} = f(x_k)$ is called the **fixed point iteration** of $f(x)$ (with initial value $x_0 = a$).

In the example above, $a = 1$ and $f(x) = \sqrt{2 + x}$. And it looks like $x_k$ is converging to $2$. How do we prove $x_k \to 2$ as $k \to \infty$?

Consider a continuous function $f(x)$ and initial value $x_0 = a$ and suppose that $x_{k+1} = f(x_k)$ converges to $L$. Then
$$
\lim_{k \to \infty} x_k = L
$$
Apply $f(x)$ to both sides of the equation
$$
f \left( \lim_{k \to \infty} x_k \right)= f(L)
$$
Since $f(x)$ is continuous
$$
\lim_{k \to \infty} f(x_k)= f(L)
$$
and so
$$
\lim_{k \to \infty} x_{k+1} = f(L)
$$
and finally
$$
L = f(L)
$$

So if we want to find the limit of fixed point iteration then we need to find $L$ that satisfies the equation $L = f(L)$.

A value $x$ is called a **fixed point** of $f(x)$ if $x = f(x)$.

How do we find fixed points? Solve $x = f(x)$ IF we can!

For the example above, we have $x = \sqrt{2 + x}$. Can we solve it?

Square both sides:
$$
x^2 = 2 + x \Rightarrow x^2 - x - 2 = 0 \Rightarrow x = 2,-1
$$
The value $x=-1$ doesn't work therefore $x=2$ is the only fixed point. We haven't proved that the sequence $x_{k+1} = \sqrt{2 + x_n}$ converges to $L=2$. We have only proved that IF the sequence converges it has to converge to 2. We need to the Fixed Point Theorem.

But what about functions where we can't solve $x = f(x)$? Like $f(x) = e^{-x}$. We can't solve $x = e^{-x}$ algebraically!

What do we do? Just show that a solution exists!

## Intermediate Value Theorem

Let $g(x)$ be a continuous function on $[a,b]$. Then for any value $K$ in between the values $g(a)$ and $g(b)$ there exists $c \in (a,b)$ such that $g(c) = K$.

In particular, we can apply IVT to show zeros of $g(x)$ exist. If $g(x)$ is continuous on $[a,b]$ and $g(a)g(b) < 0$ (ie. the sign changes from $a$ to $b$) then there exists $c \in (a,b)$ such that $g(c) = 0$.

Apply IVT to $x = e^{-x}$ to show a fixed point of $f(x)=e^{-x}$ exists.

Let $g(x) = x - e^{-x}$ and compute $g(0) = 0 - e^0 = -1$ and $g(1) = 1 - e^{-1} > 0$. Therefore the IVT implies that there is some $c \in (0,1)$ such that $g(c) = 0$ and so $c = f(c)$ and $c$ is a fixed point of $f(x) = e^{-x}$.

## Fixed Point Theorem

Let $f(x)$ be a differentiable function and let $L$ be a fixed point of $f(x)$. If $|f'(x)| < 1$ for all $x \in [L - \delta, L + \delta]$ for some $\delta > 0$ then $x_{k+1} = f(x_k)$ with $x_0 \in [L - \delta, L + \delta]$ converges to $L$.

Apply this to $f(x) = \sqrt{2 + x}$ with $x_0 = 1$. Compute
$$
f'(x) = \frac{1}{2} \frac{1}{\sqrt{2 + x}}
$$
We found a fixed point $L = 2$. Since $\sqrt{2 + x} > \sqrt{2}$ for all $x > 0$ we take the reciprocal and find
$$
\frac{1}{\sqrt{2 + x}} < \frac{1}{\sqrt{2}}
$$
for all $x$ and therefore
$$
|f'(x)| = \left| \frac{1}{2} \frac{1}{\sqrt{2 + x}} \right| < \frac{1}{2} \frac{1}{\sqrt{2}} < 1
$$
for all $x > 0$. This implies that fixed point iteration of $f(x) = \sqrt{2 + x}$ with initial value $x_0 = 1$ converges to fixed point $L=2$ (since we can take $\delta = 1$ in the fixed point theorem.
