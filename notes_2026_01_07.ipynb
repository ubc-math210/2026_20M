## LaTeX

Write mathematical notation inline using LaTeX code in single dollar signs `$ ... $`. For example, the equation of a circle of redius $R$ is $x^2 + y^2 = R^2$.

Write mathematical notataion in display mode using double dollar signs `$$ ... $$`. For example, the sum formula for the geometric series is

$$
\sum_{n=0}^N r^n = \frac{1 - r^{N+1}}{1 - r}
$$

A first order separable differential equation is of the form $y' = f(t) g(y)$. Apply separation of variables:

$$
\int \frac{dy}{g(y)} = \int f(t) dt
$$

Find the general solution of $x y' = -y^2$. Setup separation of variables:

$$
-\frac{y'}{y^2} = \frac{1}{x}
$$

Integrate:

$$
-\int\frac{dy}{y^2} = \int \frac{dx}{x}
$$

And we find:

$$
\frac{1}{y} = \ln |x| + C
$$

Solve for $y$:

$$
y(x) = \frac{1}{\ln |x| + C}
$$

The determinant of a $2 \times 2$ martix is given by the formula:

$$
\det \left( \begin{bmatrix} a & b \\ c & d \end{bmatrix} \right) = ad - bc
$$

The union of intervals $(-\infty,\frac{1}{e})$ and $(\frac{1}{e},\infty)$ is

$$
\left( -\infty,\frac{1}{e} \right) \cup \left( \frac{1}{e},\infty \right)
