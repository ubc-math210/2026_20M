## NumPy Arrays

We can create NumPy arrays using the function `np.array` and entering a list of lists of numbers:

```python
A = np.array([[1.,2.],[3.,4.]])
print(A)
```

```python
M = np.array([[-2.,1.,0.],[1.,-2.,1.],[0.,1.,-2.]])
print(M)
```

Typing out the values is tedious. We can use functions to create arrays:

```python
A = np.zeros((4,5))
print(A)
```

```python
I = np.eye(4)
print(I)
```

```python
D = np.diag([1.,2.,3.,4.])
print(D)
```

```python
O = np.ones((3,2))
print(O)
```

```python
D1 = np.diag([1.,2.,3.],-1)
print(D1)
```

```python
N = 10
A = -2*np.eye(N) + np.diag(np.ones(N-1),1) + np.diag(np.ones(N-1),-1)
print(A)
```

## scipy.linalg

There is `numpy.linalg` and `scipy.linalg`. They are similar but different. We will use `scipy.linalg`.

```python
import scipy.linalg as la
```

Use the funciton `la.solve` to compute solutions of $A \mathbf{x} = \mathbf{b}$.

```python
A = np.array([[-2.,1.,0.],[1.,-2.,1.],[0.,1.,-2.]])
b = np.array([[1.],[1.],[1.]])
print(A)
print(b)
```

```python
x = la.solve(A,b)
```
