## Example 1

Find all roots of $p(x) = x^3 + x - 1$ using fixed point iteration.

From last time, we saw that there exists a unique root in $[0,2]$. Find $f(x)$ such that a fixed point of $f(x)$ is a root of $p(x)$.

Try $f(x) = 1 - x^3$.

```python
x = 1.5
N = 5
for n in range(1,N+1):
    x = 1 - x**3
print(x)
```

We want to translate $p(x) = 0$ into $x = f(x)$. Try $f(x) = (1 - x)^{1/3}$.

```python
x = 0.5
N = 50
for n in range(1,N+1):
    x = (1 - x)**(1/3)
print(x)
```

Prove this converges using Fixed Point Theorem. Compute $f'(x)$:
$$
f'(x) = -\frac{1}{3(1 - x)^{2/3}}
$$
This might be hard to prove! Try something else.

$$x^3 + x - 1 = 0 \Rightarrow x = \frac{1 - x}{x^2}$$

```python
x = 0.5
N = 10
for n in range(1,N+1):
    x = (1 - x)/x**2
print(x)
```

$$x^3 + x - 1 = 0 \Rightarrow x = \sqrt{\frac{1 - x}{x}}$$

Prove this converges using Fixed Point Theorem. Compute $f'(x)$:
$$
f'(x) = -\frac{1}{3(1 - x)^{2/3}}
$$
We see that $|f'(x)|$ is decreasing as $x$ decreasing from 1.

Set the denominator to 1 and solve for $x$ to find where $|f'(x)| = 1$. 
$$
3(1 - x)^{3/2} = 1 \Rightarrow x = 1 - \frac{1}{3^{2/3}} 
$$

Look at $x_0 = 1 - (1/3)^{3/2}$.

```python
x = 1 - (1/3)**(3/2)
y = x**3 + x - 1
print(y)
```

Therefore $|f'(x)| < 1$ for all $x < x_0$ and so $|f'(x)| < 1$ for $x$ near $L$.

```python
x = 0.5
N = 1000
for n in range(1,N+1):
    x = (1 - x)**(1/3)
print(x)
```

Approximate all roots of $p(x) = x^5 - x - 1$ using fixed point iteration.

Compute $p'(x) = 5x^4 - 1$. Two critical points $x = \pm 1/5^{1/4}$.

```python
p = lambda x: x**5 - x - 1
c1 = -1/5**0.25
c2 = c1
print(c1,c2)
```

```python
p(c1)
```

```python
p(c2)
```
