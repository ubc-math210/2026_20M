3. Write code in the function and test often!

```python
# Version 1
def norm(x):
    n = len(x)
    terms = [x[k]**2 for k in range(n)]
    result = sum(terms)**0.5
    return result
```

```python
# Version 2
def norm(x):
    terms = [xk**2 for xk in x]
    result = sum(terms)**0.5
    return result
```

```python
c1 = norm([0,1])
print(c1,1.0)
```

```python
c2 = norm([1,1,1])
print(c2,3**0.5)
```
