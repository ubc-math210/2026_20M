## Fixed Point Iteration

Let $f(x)$ be a function and let $a$ be a real number. **Fixed point iteration** of $f(x)$ with initial value $x_0 = a$ is the recursive sequence defined by $x_0 = a, x_{k+1} = f(x_k)$ for $k = 0,1,2,\dots$

Last time: If fixed point iteration of $f(x)$ converges then it must converge to a fixed point of $f(x)$.

A helpful way to visualize fixed point iteration is to graph out the following steps:

* Start at $x_0$
* Plug in to get $y_0 = f(x_0)$ and plot the point $(x_0,y_0)$ on the graph of $y = f(x)$
* Trace the horizontal line from $(x_0,y_0)$ to the line $y=x$ to find point $(y_0,y_0)$
* Trace the vertical line from $(y_0,y_0)$ to $(y_0,f(y_0))$ on the graph $y = f(x)$
* Call $y_0$ a new name $x_1$ and write $(x_1,f(x_1))$ for the new point
* Repeat!

Apply the procedure to $f(x) = \sqrt{2 + x}$ for different initial values. It always converges!

Now apply to $f(x) = \frac{x^2}{2} + x - \frac{1}{2}$. Does fixed point iteration converge? Sometimes! Try to figure out for what initial values the iteration converges. 

This shows us that iteration converges to a fixed point $L$ when $y=f(x)$ is below $y=x$ for $L < x$, and above for $x < L$.

## Fixed Point Theorem

Let $f(x)$ be a differentiable function and let $L$ be a fixed point. If $|f'(x)| < 1$ for all $x \in [L - \delta, L + \delta]$ then $x_0 = a$, $x_{k+1} = f(x_k)$, converges to $L$ for any $a \in [L - \delta, L + \delta]$.

Proof. Let $M = \max |f'(x)|$ for $x$ near $L$. We know $M < 1$. Look at $|x_1 - L|$.
$$
\begin{align*}
| x_1 - L | &= | f(x_0) - f(L) | \\
&= \frac{|f(x_0) - L |}{| x_0 - L |} | x_0 - L | \\
&\leq M | x_0 - L | \\
\end{align*}
$$
If we repeat this calculuation we get $| x_k - L| < M^k |x_0 - L|$. Therefore $|x_k - L| \to 0$ as $k \to \infty$ and so we we conclude that $x_k \to L$ as $k \to \infty$.
