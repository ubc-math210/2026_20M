## Vectorization

Vectorization is the process of applying an operation/function to a vector by applying to every entry in the vector.

Vector addition and scalar multiplication are vectorized. Let's do some more examples.

Vectorized multiplication:

```python
v * w
```

Vectorized division:

```python
v/w
```

Vectorized exponentiation:

```python
v**w
```

```python
w**v
```

The vectorization of a function $f(x)$ is simply applying $f(x)$ to every entry in a vector $\mathbf{x}$:

$$
f(\mathbf{x}) = \begin{bmatrix} f(x_0) & f(x_1) & \cdots & f(x_N)\end{bmatrix}
$$

## Plotting Examples

Plot the solution of the equation

$$
y'' + 3y' + 2y = 1 \ , \ y(0) = 0.5 \ , \ \ y'(0) = 1
$$

Roots of the characteristic polynomial are $r=-1$ and $r=-2$. Solution to the homogeneous equation is:

$$
y_h(x) = C_1 e^{-x} + C_2 e^{-2x}
$$

Find a particular solution of the form $y_p(x) = A$. Plug in and solve $A=1/2$. General solution is:

$$
y(x) = \frac{1}{2} + C_1 e^{-x} + C_2 e^{-2x} \ , \ \ C_1,C_2 \in \mathbb{R}
$$

Plug in initial conditions:

$$
\frac{1}{2} + C_1 + C_2 = 0.5 \ , \ \ -C_1 -2 C_2 = 1
$$

And we find $C_1 = 1$ and $C_2 = -1$

```python
x = np.linspace(0,1,100)
y = 1/2 + np.exp(-x) - np.exp(-2*x)
plt.plot(x,y), plt.ylim([0,1]), plt.axis('equal'), plt.grid(True)
plt.show()
```

## Example 2

Plot the solution of

$$
y'' + 2y' + 2y = 0 \ , \ \ y(0) = y_0 \ , \ \ y'(0) = 0
$$

for different values of $y_0$.

Roots of characteristic polynomial are $r_1 = -1 + i$ and $r_2 = -1 - i$.

The solution of the homogeneous equation is:

$$
y(x) = C_1 e^{-x} \cos(x) + C_2 e^{-x} \sin(x)
$$

Note that $y(0) = C_1$. The derivative is

$$
y'(x) = -C_1 e^{-x} \cos(x) - C_1 e^{-x} \sin(x) - C_2 e^{-x} \sin(x) + C_2 e^{-x} \cos(x)
$$

$$
y'(0) = -C_1 + C_2 = 0 \Rightarrow C_1 = C_2
$$

```python
x = np.linspace(0,5,200)
for y0 in np.linspace(0,1,11):
    y = y0*np.exp(-x)*np.cos(x) + y0*np.exp(-x)*np.sin(x)
    plt.plot(x,y,'b')
plt.show()
```
