## Python Packages and Modules

A Python module is a text file (with extension `.py`) that contains Python code. We import modules with `import` keyword. For example, let's make our own module called `roots` which contains all our rooting finding functions.

````python
import roots
````

We access functions and objects inside a module or package using the dot syntax.

````python
f = lambda x: x**2 - 2
Df = lambda x: 2*x
x0 = 1.5
r = roots.newton(f,Df,x0,1e-8,20)
print(r)
````

````python
f = lambda x: 1/(1 + x)
x0 = 1.
r = roots.fiteration(f,x0,20)
print(r)
````

A Python package is a directory of Python files containing Python code. We import using the same `import` keyword and access functions, objects, subpackages, etc. inside the package using the dot syntax.

### Example

````python
x = np.linspace(-2,2,200)
y = np.exp(-x**2)*np.cos(10*x)
plt.plot(x,y), plt.grid(True), plt.title("Decaying Sine Wave")
plt.show()
````

### Vectorization

The function `np.linspace(a,b,N+1)` creates an array of $N$ equally spaced from `a` to `b` (inclusive).

````python
x = np.linspace(0,1,11)
print(x)
````

````python
type(x)
````

If $f(x)$ is a function and
$$
\mathbf{x} = \begin{bmatrix} x_0 & x_1 & \cdots & x_N \end{bmatrix}
$$
is a vector then $f(\mathbf{x})$ is the vector
$$
f(\mathbf{x}) = \begin{bmatrix} f(x_0) & f(x_1) & \cdots & f(x_N) \end{bmatrix}
$$
This is the vectorization of the function $f(x)$. In particular in the code above the expression
```
y = np.exp(-x**2)*np.cos(10*x)
```
is given by the vectorization formula
$$
\mathbf{y} = \begin{bmatrix} e^{-x_0^2} \cos(10x_0) & e^{-x_1^2} \cos(10x_1) & \cdots & e^{-x_N^2} \cos(10x_N) \end{bmatrix}
$$

### Example

Write a for loop to plot $y = e^{-kx}$ on $[0,4]$ for $k=1,2,\dots,10$.

````python
x = np.linspace(0,4,101)
for k in range(1,11):
    y = np.exp(-k*x)
    plt.plot(x,y,'b')
plt.show()
````
