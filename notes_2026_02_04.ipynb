## Secant Method

Consider a continuous function $f(x)$ on $[a,b]$ such that $f(a) f(b) < 0$. A **secant line** of $f(x)$ is a straight line connecting two points on the graph $y = f(x)$.

In the bisection method, we don't use any information about the shape of $f(x)$ when computing the next interval using the midpoint. In the secant method, we use the $x$-intercept of the secant line and apply the same algorithm as the bisection method.

The equation of the line connecting points $(a,f(a))$ and $(b,f(b))$ is

$$
y = \frac{f(b) - f(a)}{b - a}(x - a) + f(a)
$$

Solve for the $x$-intercept

$$
0 = \frac{f(b) - f(a)}{b - a}(x - a) + f(a)
\Rightarrow
x = a - f(a)\frac{b - a}{f(b) - f(a)}
$$

Write a function called `secant` which takes function `f`, numbers `a` and `b`, integer `N`, and returns the $x$-intercept of the Nth interval from secant method.

```python
def secant(f,a,b,N):
    if f(a)*f(b) >= 0:
        print("Initial interval does not satisfy f(a)f(b) < 0.")
        return None
    a_n = a
    b_n = b
    for n in range(1,N+1):
        f_a_n = f(a_n)
        f_b_n = f(b_n)
        m_n = a_n - f_a_n*(b_n - a_n)/(f_b_n - f_a_n)
        f_m_n = f(m_n)
        if f_a_n*f_m_n < 0:
            b_n = m_n
        elif f_b_n*f_m_n < 0:
            a_n = m_n
        elif f_m_n == 0:
            print("Found exact solution.")
            return m_n
        else:
            print("Secant method fails.")
            return None
    return a_n - f(a_n)*(b_n - a_n)/(f(b_n) - f(a_n))
```

```python
f = lambda x: x**2 - 2
a = 1
b = 2
N = 20
r = secant(f,a,b,N)
print(r)
```

Compare with the bisection method:

```python
def bisection(f,a,b,N):
    if f(a)*f(b) >= 0:
        print("Initial interval does not satisfy f(a)f(b) < 0.")
        return None
    a_n = a
    b_n = b
    for n in range(1,N+1):
        m_n = (a_n + b_n)/2
        f_m_n = f(m_n)
        if f(a_n)*f_m_n < 0:
            b_n = m_n
        elif f(b_n)*f_m_n < 0:
            a_n = m_n
        elif f_m_n == 0:
            print("Found exact solution.")
            return m_n
        else:
            print("Bisection method fails.")
            return None
    return (a_n + b_n)/2
```

```python
f = lambda x: x**2 - 7
a = 2
b = 3
N = 3
r1 = secant(f,a,b,N)
r2 = bisection(f,a,b,N)
print("Secant:",r1,abs(7**0.5 - r1))
print("Bisection:",r2,abs(7**0.5 - r2))
```
