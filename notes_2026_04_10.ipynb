```python
import numpy as np
import scipy.linalg as la
```

```python
A = np.array([[1.,1.],[1.,-2.]])
print(A)
```

```python
D,P = la.eig(A)
print(D)
print(P)
```

```python
x = np.array([[1.],[0.]])
for k in range(11):
    x = A@x
    #x = x/np.max(np.abs(x))
    x = x/la.norm(x)
print(x)
```

```python
eval1 = np.sum(x*(A@x))/np.sum(x*x)
```
